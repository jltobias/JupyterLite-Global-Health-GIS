# Public data sources, citation, and ethics checklist

Use this notebook as a pre-flight checklist before adding real public health data to a public GitHub repository.

## Suggested data sources

- IHME / GHDx / GBD Results for burden estimates, with the correct citation and user agreement.
- WHO Global Health Observatory for indicators that are designed for public API access.
- World Bank DataBank or API for demographic, development, and health-system covariates.
- Natural Earth or other public boundary sources for low-resolution country teaching maps.
- Humanitarian Data Exchange for selected operational layers, checking license and update cadence.

## Do-not-commit checklist

- No line lists, individual records, names, IDs, addresses, coordinates from patients, or facility-sensitive operational data.
- No data with unclear redistribution terms.
- No API keys, tokens, private URLs, credentials, or internal server names.
- No full raw downloads that make the Lite site slow or impossible to load.
- No charts or maps without a visible data source, access date, and caveat about uncertainty or limitations.
