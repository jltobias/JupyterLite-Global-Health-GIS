# JupyterGIS: global health point layers

This notebook opens a global health burden point layer with the JupyterGIS Python API. In a full JupyterLab deployment, JupyterGIS supports collaborative workflows. In JupyterLite, the map runs locally in the browser, so real-time collaboration is intentionally not available.

The bundled data are synthetic. Replace them with a licensed IHME/GHDx export processed by the pipeline notebook.

```python
from pathlib import Path
import json
import pandas as pd

geojson_path = Path("data/gbd_country_points_2021.geojson")
if not geojson_path.exists():
    geojson_path = Path("data/gbd_country_points_generated.geojson")

print("GeoJSON:", geojson_path)
data = json.loads(geojson_path.read_text(encoding="utf-8"))
print("Features:", len(data.get("features", [])))
pd.DataFrame([f["properties"] for f in data["features"]]).head()
```

```python
# Primary path: JupyterGIS.
# This cell is intentionally simple so it remains robust across JupyterGIS releases.
try:
    from jupytergis import GISDocument
    doc = GISDocument(latitude=12, longitude=8, zoom=1.2)
    doc.add_geojson_layer(path=str(geojson_path), name="Global health burden points: synthetic demo")
    doc
except Exception as exc:
    print("JupyterGIS was not available in this kernel or failed to initialize.")
    print("Reason:", repr(exc))
    print("The data are still available as GeoJSON and can be opened in GeoLibre or another GIS tool.")
```

```python
# Optional: try a heatmap layer weighted by the 'val' field.
try:
    from jupytergis import GISDocument
    heat_doc = GISDocument(latitude=12, longitude=8, zoom=1.2)
    heat_doc.add_heatmap_layer(feature="val", path=str(geojson_path), name="Synthetic burden heatmap")
    heat_doc
except Exception as exc:
    print("Heatmap example did not run in this environment:", repr(exc))
```
