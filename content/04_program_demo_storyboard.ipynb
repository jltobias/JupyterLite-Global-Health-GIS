# Program demonstration storyboard

This notebook turns the sample dataset into reusable talking points for global health program demonstrations. It is designed for non-specialist audiences: program managers, country teams, informatics staff, and training cohorts.

```python
import pandas as pd
import matplotlib.pyplot as plt
from pathlib import Path

df = pd.read_csv(Path("data/sample_gbd_like_burden.csv"))
latest = df[df["year"] == df["year"].max()].copy()
latest.head()
```

```python
# Example 1: burden transition by region.
regional = (
    latest
    .groupby(["region_name", "cause_group"], as_index=False)
    .agg(mean_demo_rate=("val", "mean"))
    .sort_values(["region_name", "mean_demo_rate"], ascending=[True, False])
)
regional.head(10)
```

```python
# A compact figure that works in JupyterLite.
pivot = regional.pivot_table(index="region_name", columns="cause_group", values="mean_demo_rate", aggfunc="mean").fillna(0)
ax = pivot.plot(kind="barh", figsize=(9, 6))
ax.set_xlabel("Mean synthetic DALY rate per 100,000")
ax.set_ylabel("")
ax.set_title("Synthetic burden profile by region and cause group")
plt.tight_layout()
plt.show()
```
