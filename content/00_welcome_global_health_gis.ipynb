# Welcome: Global Health GIS in the Browser

This notebook introduces the repository and confirms that the browser-based Python environment can read the bundled demo data.

**Important data note:** the bundled `sample_gbd_like_burden.csv` file is artificial. It mimics the column structure of a Global Burden of Disease style export so that the notebooks run immediately in JupyterLite, but the values are **not** IHME estimates and must not be interpreted as health evidence.

Suggested live-demo flow:

1. Open this notebook to verify the data and browser environment.
2. Run `01_gbd_data_pipeline_template.ipynb` to process a GBD-style CSV into GeoJSON and a GeoLibre project.
3. Run `02_jupytergis_global_health_points.ipynb` to display the data with JupyterGIS.
4. Run `03_geolibre_project_launcher.ipynb` to open the same project in GeoLibre Web.

```python
from pathlib import Path
import pandas as pd

data_path = Path("data/sample_gbd_like_burden.csv")
df = pd.read_csv(data_path)
print(f"Rows: {len(df):,}")
print(f"Countries: {df['iso3'].nunique():,}")
print(f"Years: {sorted(df['year'].unique())}")
df.head()
```

```python
# Summarize the latest year of the demo data.
latest = df[df["year"] == df["year"].max()].copy()
summary = (
    latest
    .groupby("cause_name", as_index=False)
    .agg(mean_demo_rate=("val", "mean"), countries=("iso3", "nunique"))
    .sort_values("mean_demo_rate", ascending=False)
)
summary
```

```python
# A small browser-friendly chart. This uses synthetic values only.
import matplotlib.pyplot as plt

ax = summary.plot.bar(x="cause_name", y="mean_demo_rate", legend=False, figsize=(8, 4))
ax.set_ylabel("Mean synthetic DALY rate per 100,000")
ax.set_xlabel("")
ax.set_title("Synthetic demo burden profile, latest year")
plt.xticks(rotation=30, ha="right")
plt.tight_layout()
plt.show()
```
