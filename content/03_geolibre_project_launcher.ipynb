# GeoLibre project launcher

GeoLibre can open `.geolibre.json` projects in the browser. This notebook previews the bundled project file and builds launch links.

For a public GitHub repository, GeoLibre Web can open the raw project URL directly. For local testing before you push the repository, open GeoLibre Web and use the app's open/import controls to select `content/projects/global_health_demo.geolibre.json` from your local checkout.

```python
from pathlib import Path
import json
from urllib.parse import quote
from IPython.display import HTML, display, IFrame

project_path = Path("projects/global_health_demo.geolibre.json")
project = json.loads(project_path.read_text(encoding="utf-8"))
print(project["name"])
print("Layers:", [layer["name"] for layer in project["layers"]])
```

```python
# Update these when you publish the repository under your GitHub profile.
GITHUB_USER = "jltobias"
REPO_NAME = "jupyterlite-global-health-gis"
BRANCH = "main"
raw_url = f"https://raw.githubusercontent.com/{GITHUB_USER}/{REPO_NAME}/{BRANCH}/content/projects/global_health_demo.geolibre.json"
geolibre_url = "https://web.geolibre.app/?url=" + quote(raw_url, safe="") + "&layout=compact"
print("Raw project URL:")
print(raw_url)
print("GeoLibre launch URL:")
print(geolibre_url)
display(HTML(f'<p><a href="{geolibre_url}" target="_blank" style="font-size: 120%;">Open the project in GeoLibre Web</a></p>'))
```

```python
# Embedded view. Some browsers, privacy settings, or GitHub raw URL policies may block iframes;
# if so, use the link printed above.
IFrame(src=geolibre_url, width="100%", height=650)
```
